# Cleaning Data

```python
import pandas as pd
import numpy as np

superstoreData = pd.read_csv('../data/Superstore.csv', encoding='latin1')
df = superstoreData.copy()
df.head()
```

```python
df.isnull().sum()
```

## Observation

### There are none missing value

```python
duplicate_mask = df.duplicated()
num_duplicates = duplicate_mask.sum()
print(f"Number of duplicate rows: {num_duplicates}")

df = df.drop_duplicates().reset_index(drop=True)
print(f"New shape after removing duplicates: {df.shape}")
```

## Observation

### After observation there are no duplicate data.

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
df["order_date"].dtype, df["ship_date"].dtype
df["ship_date"].head()
df["order_date"].head()
```

```python
# Convert column names to lowercase with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Convert city names to lowercase
df['city'] = df['city'].str.lower()
df['state'] = df['state'].str.lower()
df['country'] = df['country'].str.lower()
df.head()
```

```python
df["sales"].describe()
```

## Some question to help with more cleaning after.

### Can Sales be negative? No sales can not be negative because there is the price of the item.
### Can Quantity be 0? No quantity can not be 0 because it means that this item was not bought from the client and we should not include this. 
### Can Discount exceed 1? No if it cross 0 then the item will be sold for free 
### Can Profit be negative? Yes profit can be negative because of transport cost and discount 

```python
df["ship_time"] = (df["ship_date"] - df["order_date"]).dt.days

for num in df['ship_time']:
    if num < 0:
        print("There are negative values in the ship_time column.")
        break
    else:
        print("There are no negative values in the ship_time column.")
        break

df['ship_time'].describe()
```

#### Observation
##### There are no negative shippment time with a max 7 days shippment mean 4 days
