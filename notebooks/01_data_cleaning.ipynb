```python
from IPython.display import display, Markdown

def infer_col_type(s: pd.Series, parse_thresh: float = 0.8) -> str:
    if pd.api.types.is_bool_dtype(s):
        return "bool"
    if pd.api.types.is_numeric_dtype(s):
        return "numeric"
    if pd.api.types.is_datetime64_any_dtype(s):
        return "datetime"
    if pd.api.types.is_timedelta64_dtype(s):
        return "timedelta"
    if pd.api.types.is_categorical_dtype(s):
        return "category"
    if pd.api.types.is_object_dtype(s):
        # try parsing to datetime to detect date-like strings
        parsed = pd.to_datetime(s, errors="coerce", infer_datetime_format=True)
        frac_parsed = parsed.notna().mean() if len(parsed) > 0 else 0.0
        if frac_parsed >= parse_thresh:
            return f"datetime (parsed {frac_parsed:.0%})"
        return "string/object"
    return str(s.dtype)

rows = []
for col in df.columns:
    pandas_dtype = df[col].dtype
    inferred = infer_col_type(df[col])
    rows.append((col, str(pandas_dtype), inferred))

md = "| Column | pandas_dtype | inferred_type |\n|---|---:|---|\n"
for col, pdtype, inf in rows:
    md += f"| {col} | {pdtype} | {inf} |\n"

display(Markdown(md))
```
